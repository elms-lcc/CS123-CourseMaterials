```python
# @title Imports
import matplotlib
import scipy
import numpy as np
import pandas as pd
```

```python
# @title Starting data (feel free to try modifying)

input_data = [(0.0, False,),
              (2.0, False,),
              (3.0, False,),
              (4.0, True,),
              (5.0, False,),
              (6.0, True,),
              (7.0, True,),
              (8.0, True,),
              (9.5, True,),
              ]
data = pd.DataFrame(input_data, index=None, columns=['Hours','Pass'])

# display data
data
```

# Code to fit curve

```python
# @title Define logistic function
# @markdown We will fit the hours to the logistic function $ f(h) = \frac{1}{1 + e^{-(w\cdot h + b)}}$
def logistic_regression(x, w, b):
  return 1. / (1. + np.exp(-(w * x + b)))
```

```python
# @title Fit hours and passing data to logistic function
p, pc = scipy.optimize.curve_fit(logistic_regression, data.Hours, data.Pass)
```

```python
# @title Plot fit and data
# generate curve fitted
xi = np.linspace(0,10,100)
yi = 1. / (1. + np.exp(-(p[0]*xi + p[1])))
# plot
matplotlib.pyplot.plot(xi, yi)
_ = matplotlib.pyplot.scatter(data.Hours, data.Pass)
```

```python
# @title Fit parameters
print(f"w={p[0]}")
print(f"b={p[1]}")
```
